```python
import re
```

```python
#match(pattern,string):It checks whether the string is starts with pattern or not 
string="The sky is blue and ocean is blue"
res=re.match("The",string)
print(res)
print(res.span()) #it returns the matched pattern positions
print(res.group()) #it returns the matched result 
```

```python
#match
string="The sky is blue and ocean is blue"
res=re.match("sky",string)
print(res)
print(res.span()) #it returns the matched pattern positions
print(res.group()) #it returns the matched result 
```

```python
#match
#to remove error use try and except
try:
    string="The sky is blue and ocean is blue"
    res=re.match("sky",string)
    print(res)
    print(res.span()) #it returns the matched pattern positions
    print(res.group()) #it returns the matched result 
except Exception as e:
    print("There is no matched pattern")
```

```python
#search(patt,string):It checks the entire string and it return the first matched pattern.
try:
    string="The sky is blue and ocean is blue"
    res=re.search("sky",string)
    print(res)
    print(res.span()) #it returns the matched pattern positions
    print(res.group()) #it returns the matched result 
except Exception as e:
    print("There is no matched pattern")
```

```python
#search
try:
    string="The sky is blue and ocean is blue"
    res=re.search("z",string)
    print(res)
    print(res.span()) #it returns the matched pattern positions
    print(res.group()) #it returns the matched result 
except Exception as e:
    print("There is no matched pattern")
```

```python
#search
try:
    string="The sky is blue and ocean is blue"
    res=re.search("blue",string)
    print(res)
    print(res.span()) #it returns the matched pattern positions
    print(res.group()) #it returns the matched result 
except Exception as e:
    print("There is no matched pattern")
```

```python
#findall(pat,string):It returns the all matched patterns as list
try:
    string="The sky is blue and ocean is blue"
    res=re.findall("blue",string)
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
#findall
try:
    string="The sky is blue and ocean is blue"
    res=re.findall("[sbco]",string)
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
#split(pat,string):It splits the string wherever the pattern is matched.
try:
    string="The sky is blue and ocean is blue"
    res=re.split("blue",string)
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
#sub(old_pat,new_pat,string):it replaces old pattern with new pattern.
try:
    string="The sky is blue and ocean is blue"
    res=re.sub("blue","Red",string)
    print(res)
except Exception as e:
    print("There is no matched pattern")
```

```python
#validate mobile number
import re
try:
    mobile=input()
    res=re.fullmatch("^[6-9][0-9]{9}",mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is invalid")
except:
    print("There is no matched pattern")
```

```python
#validate aadhar number
try:
    aadhar=input()
    res=re.fullmatch(r"^[2-9][0-9]{3}\s?[0-9]{4}\s?[0-9]{4}",aadhar)
    print(res)
    if res:
        print(f"{aadhar} is valid")
    else:
        print(f"{aadhar} is invalid")
except:
    print("There is no matched pattern")
```

```python
#Email validation
try:
    email=input()
    res=re.fullmatch(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9]+\.[a-z]{2,3}$",email)
    #print(res)
    if res:
        print(f"{email} is valid")
    else:
        print(f"{email} is invalid")
except:
    print("There is no matched pattern")
```

```python
#Date validation
try:
    date=input()
    res=re.fullmatch(r"^(0[1-9]|[12][0-9]|3[01])/(0[1-9]|1[0-2])/(\d{4})",date)
    #print(res)
    if res:
        print(f"{date} is valid")
    else:
        print(f"{date} is invalid")
except:
    print("There is no matched pattern")
```

```python
#Date validation
try:
    date=input()
    res=re.fullmatch(r"^(0?[1-9]|[12][0-9]|3[01])[/-](0?[1-9]|1[0-2])[/-](\d{4})",date)
    #print(res)
    if res:
        print(f"{date} is valid")
    else:
        print(f"{date} is invalid")
except:
    print("There is no matched pattern")
```

```python
#password validation
import re
try:
    password = input()
    res = re.fullmatch(r"^(?=.*[a-z])(?=.*[A-Z])(?=.*\d)(?=.*[@#$%^&+=!]).{8,}$", password)
    #print(res)
    if res:
        print(f"{password} is valid")
    else:
        print(f"{password} is invalid")
except:
    print("There is no matched pattern")
```
