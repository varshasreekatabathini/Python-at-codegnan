- `File`:The file is used to store the data in electronic format in system.
- `File operations`:
  1. Opening a file---open()
  2. Reading a file---read()
  3. Writing into a file---write()
  4. Closing a file---close()
1. `Opening a file`:To open a file by using open() in that,we need to pass file name and file opening mode.<br>
   `syntax`:<br>
   `        file_obj=open(file_name,mode)`<br>
   `           #read/write operations`<br>
   `           #close file`<br>
   - `file opening modes`:
     - r->(read):opening file in read mode.if file not exist,it raises FileNotFoundError.
     - w->(write):if file not exist,it creates a new file.otherwise,it erase the previous content.
     - a->(append):opening a file in append mode.if file not exist it creates a new file,otherwise it add content to the previous content.
     - x->(create):Used to create new file
     - b->(binary):This mode is used to perform operations on binary files.for writing wb mode.for reading rb mode.
     - r+->read+write
     - w+->write+read
     - a+->append+read()
2. `Reading Content from a file`:To perform the read operation on file for that we need to open file in 'r' mode and then we can perform folowing operations.
   - read():It return entire file content as single string.
   - readline():This method return first line.
   - readlines():It returns list of lines.
3. `Writing into a file`:If you open a file in write mode,we can perform following operations.
   - write(string):It add string into a file.
   - writelines(list_of_lines):It adds all list of lines to a file.
4. `Closing a file`:After perform read/write operation on a file,we need to close the file by using close(),otherwise data loss happened.
