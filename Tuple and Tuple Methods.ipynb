### Characteristics

```python
#ordered collection
t=(1,2,3,[10,20,30],25)
print(t)
```

```python
#immutable
t=(1,2,3,[10,20,30],25)
t[1]=100
print(t)
```

```python
#immutable: After operation it doesn't change the length of object and doesn't change remaining elements position
t=(1,2,3,[10,20,30],25)
t[3].append(100)
print(t)
```
