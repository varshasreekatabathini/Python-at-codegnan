# Exception Handling
- Exception is nothing but it is an error.whenever exception is occurs it stops the pogram execution and it returns type of exception and followed by exception message.
- `Types of errors`:
  1. compile time errors or syntax error
  2. runtime error
  3. logical error
1. `Syntax Error`:The syntax errors raise during the compilation time of the program.Thpse errors call it as syntax error.
   - Ex:Indentation error,syntax error
2. `Runtime Error`:The runtime error raises during the runtime of a program.Those errors called as runtime errors.
   - Ex:KeyError,ValueError,ZeroDivisionError,IndexError,TypeError,etc...
   - we cannot handle syntax errors,but we can handle the runtime errors by using exception handling blocks(i.e,try,except block).
   - `Exception handling blocks`:<br>
   - `syntax:`<br>
   - `        try:`
   - `           #block code that might contains exception.`
   - `        except:`
   - `           #exception handling block`
   - `        else:[optional]`
   - `           #This block executes there is no exception raised in try block`
   - `        finally:[optional]`
   - `           #This block always executes every exception raised or not`

```python
#try block with exception
try:
    list=[1,2,3]
    print(list[10])
    print("hi")
except:
    print("Something wrong in try block")
else:
    print("No exception raised in try block")
finally:
    print("Program done")
```

```python
#There is no exception in try block
try:
    list=[1,2,3]
    print(list[2])
    print("hi")
except:
    print("Something wrong in try block")
else:
    print("No exception raised in try block")
finally:
    print("Program done")
```

```python
#It executes the specific exception block when specified exception raised otherwise,it executes common except block..
try:
    a=10
    b=a.append(1)
    c=a/b
    print(c)
except ZeroDivisionError as e:
    print(f"Zero division not possible:{e}")
except ValueError as e:
    print(f"Value Error raised in try block:{e}")
except Exception as E:                          #common exception block with exception class
    print(f"Something wrong in try block:{E}")
```
