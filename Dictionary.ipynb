```python
#Dictionary values access by using keys
#Updating dictionary values
#adding new items to dictionary
details={"Name":"Banty", "Course":"Python", "Batch":57, "age":20}
print(details['Course'])
#print(details['Location']) #KeyError: 'Location'
#updating dictionary values
details['age']=25
print(details['age'])
#adding new items
details['location']="Hyderabad"
print(details)
details['age']=details['age']+1
print(details['age'])
```

### Dictionary Methods
-get(key,default-value:Optional):It returns the value if key is exists in dictionary,otherwise it retuns default value

```python
details={"age":25,"Name":"Banty", "Course":"Python", "Batch":57, "age":20}
print(details['Name'])
print(details.get('Name'))
print(details.get('Marks'))
print(details.get('Marks', "Marks not available"))
```

```python
details={"age":25,"Name":"Banty", "Course":"Python", "Batch":57, "age":20}
print(details['Marks'])
```

- update(new_dictionary):This method updates the values if new_dictionary keys already exists in Dictionary,otherwise it adds new items to dictionary

```python
details={"age":25,"Name":"Banty", "Course":"Python", "Batch":57}
details.update({'Batch':58,'Course':'Java','Location':'Hyd'})
print(details)
```

- keys():This method returns the list of dictionary keys
- values():This method returns the list of dictionary values
- items():This method returns the list of tuple of key_value pairs 

```python
details={'age': 25, 'Name': 'Banty', 'Course': 'Python', 'Batch': 58, 'Location': 'Hyd'}
keys=details.keys()
print(type(keys))
print(keys)
print(details.values())
print(details.items())
```

- pop(item):It return and removes the last item as tuple
- pop(key):This method removes the specified key item and return the specified keys value,otherwise KeyError

```python
details={'age': 25, 'Name': 'Banty', 'Course': 'Python', 'Batch': 58, 'Location': 'Hyd'}
item=details.popitem()
print(item)
print(details)
val=details.pop('Name')
print(val)
print(details)
```
