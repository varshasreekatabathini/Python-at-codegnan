```python
#check which one is even and which one is odd
lst=list(map(int,input().split()))
for num in lst:
    if(num%2==0):
        print(f"{num} is even")
    else:
        print(f"{num} is odd")
else:
    print("Program Done")
```

```python
lst=map(int,input().split())
for num in lst:
    if(num%2==0):
        print(f"{num} * {num}=",num*num)
    else:
        print(f"{num} + {1}=",num+1)
```

```python
lst=list(map(int,input().split()))
for i in lst:
    if i%3==0 and i%5==0:
        print("FizzBuzz")
    elif i%3==0:
        print("Fizz")
    elif i%5==0:
        print("Buzz")
    else:
        print(f"{i}")
```

```python
lst=list(map(int,input().split()))
res=[]            #to store result in list
for i in lst:
    if i%3==0 and i%5==0:
        res.append("FizzBuzz")      #to add in list
    elif i%3==0:
        res.append("Fizz")
    elif i%5==0:
        res.append("Buzz")
    else:
        res.append(f"{i}")
print(res)
```

```python
#prime number
n=int(input())
for i in range(2,n//2+1):
    if n%i==0:
        print(f"{n} not prime")
else:
    print(f"{n} is prime")
```

```python
#Nested Loops
n=int(input())
for i in range(2,n):
    for j in range(2,i//2+1):
        if i%j==0:
            print(f"{i} not prime")
    else:
        print(f"{i} is prime")
```
