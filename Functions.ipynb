#### Scopes of Variable
- 1.local scope
- 2.enclosing scope
- 3.global scope
- 4.buit-in scope


```python
x=10           #global scope variable
def outer():
    y=15          #enclosing scope variable
    def inner():
        z=20           #local scope variable
        print(z,y,x)        #print()-buit-in scope
    inner()
outer()
```

```python
#using global scope variable in local scope
x=10
def fun():
    y=x+1
    print("Local scope variable:",y)
fun()
print("Global scope variable:",x)
```

```python
#Updating the global scope variable value in local scope
#cannot update global scope variable in local scope directly.
x=10
def fun():
    x=x+1
    print("Local scope variable:",x)
fun()
print("Global scope variable:",x)
```

```python
#Updating the global scope variable value in local scope bby using "global" keyword.
x=10
def fun():
    global x
    x=x+1
    print("Local scope variable:",x)
fun()
print("Global scope variable:",x)
```

```python
#Updating enclosing scope variable in local scope
x=10
def outer():
    y=15
    def inner():
        z=30
        y=y+z
        print(z,y,x)
    inner()
outer()
```

```python
#Updating enclosing scope variable in local scope by using nonlocal
x=10
def outer():
    y=15
    def inner():
        nonlocal y
        z=30
        y=y+z
        print(z,y,x)
    inner()
    print("Enclosing scope variable:",y)
outer()
```

```python
x=10
def outer():
    x=15
    x=x+2
    print("Local scope variable:",x)
outer()
print("Global scope variable:",x)
```

```python
s1="CODE"
s2="GNAN"
n=len(s1)
s=""
for i in range(n):
    res=abs(ord(s1[i])-ord(s2[i]))
    if res==0:
        s+=s1[i]
    else:
        s+=chr(64+res)
print(s)
```

```python
s1="CODE"
s2="GNAN"               
n=len(s1)
a=""
b=""
for i in range(n):
    if s1[i] !=s2[i]:
        if ord(s1[i])>ord(s2[i]):
            a+=s1[i]
            b+=s2[i]
        else:
            a+=s2[i]
            b+=s1[i]
print(a)
print(b)
```

```python
#map(fun, iterable):It applies the functionality on each element in iterable
#filter(fun,iterable): It filters the iterables,it returns the values which is not None.
#Find the squares of list of integers
def square(x):
    return x**2
lst=[1,2,3,4,5]
print(list(map(square, lst)))
```

```python
#Filter the even numbers from list of integers
def even(x):
    if x%2==0:
        return x
arr=[1,2,3,4,5]
print(list(filter(even,arr)))
print(list(map(even,arr)))
```

```python
def sub(z):
    x,y=z
    return x+y
lst=[(1,2),(3,4),(5,6)]
print(list(map(sub, lst)))
```

```python
#Find the target element position in list
arr=[10,20,30,40,50]
target=30
for ind,val in enumerate(arr):
    if target==val:
        print(ind)
        break
else:
    print(-1)
```
