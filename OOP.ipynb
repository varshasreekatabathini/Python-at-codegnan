### Encapsulation:
   - Encapsulation means,wrapping the methods and attributes in single unit that one is class.The main of encapsulation is to restrict the direct access of attributes and methods.
   - This type of restriction can implement by using `Access Modifiers`
   - Types of Access Modifiers
       1. Public     -without underscore
       2. Protected  -starts with single underscore
       3. Private    -starts with double underscore
   - In python we don't have any keywords to represent it public,protected and private, but by using underscore(_) can define.
   - Python dosen't secure high security like java.Here we can access the protected memebers and private members in outside the class by using following ways:
   - To access the protected member-(obj._protectedmembername)
   - To access the private member by using Name Mangling-(obj._classname__privatemembername)
   - The advantage of encapsulation is can provide the security of the data as well as functionalities...
     1. Public:The public members(i.e attributes+method) dosen't starts with underscore.This public memebers can access any where in the program
     2. Protected:The protected members(i.e attributes+method) starts with single underscore(_).This memebers can access in inside the class and in subclass also.
     3. Private:The private members(i.e attributes+method) starts with double underscore(__).This memebers can access in only inside the class.

  |   Access Modifiers   |   Inside class   |   Sub class    |   Outside class        | <br>
         public                  yes               yes                 yes               <br>
         protected                 yes               yes              no                                                       <br>
         private                   yes               no               no                                                            <br>

```python
class Bank:
    def __init__(self,name,account,password,balance):
        self.name=name
        self.account=account
        self.password=password
        self._balance=balance
    def __total_users(self):           #getter method
        return "Total users : 100"
    def get_balance(self):           #getter method
        return f"Current Balance is:{self._balance}"
    def deposit(self, deposit_amount):      #setter method
        self._balance+=deposit_amount
        return f" {deposit_amount} Deposited Successfully"
u1=Bank("Lahari", 12, "Lahari@02", 5000)
print(u1.name)
print(u1.get_balance())
print(u1.deposit(5000))
print(u1.get_balance())
print(u1._balance)         #protected memebers access in outside the class
print(u1._Bank__total_users())     #Name Mangling:syntax:obj._classname__attribute)
```

### Inheritance

- In python, A class get properties from another class this process is called as `Inheritance`.
  - A child class is inherited from parent class.
  - Types of Inheritance
    1. single inheritance
    2. mutli-level inheritance
    3. multiple inheritance
    4. hierarchical inheritance
    5. hybrid inheritance
1. `Single Inheritance`:A child class get properties from single parent class.This is called as `Single Inheritance`.<br>
   `Example`: Fruits<--Mango <br>
   `Syntax:`
   `        class parent_class_name:`<br>
   `                #attributes`<br>
   `                #method`    <br>
   `        class child_class_name(parent_class_name):`<br>
   `                #attributes` <br>
   `                #methods`    <br>
2. `Multi-Level Inheritance`:A class(i.e child class) get properties from another class(i.e parent class) that class get properties from some another class(i.e grand parent).<br>
   `Example`:Employee<--Team Lead<--Manager<--HR  <br>
3. `Multiple Inheritance`:A single child class get properties from multiple parent classes.This is called as `Multiple Inheritance`.<br>
   `Example:`swiggy,zomato,blinkit<---delivery boy
4. `Hierarchical Inheritance`:Mutiple child classes get properties from single parent class.<br>
   `Example:`rivers-->krishna,godavari,yamuna
6. `Hybrid Inheritance`:The combination of two or more types of inheritance is called as Hybrid Inheritance.

### Polymorphism

- In Python polymorphism `poly` means `many` and `morphism` means `forms`.
- In polymorphism single function, it shows differnet behaviour based on input context.
- In python polymorphism can achieve through following ways:
  1. Polymorphism through operators
  2. Polymorphism through functions
  3. Polymorphism through methods
     - Runtime polymorphism(Method Overriding)
     - Compile time polymorphism(Method Overloading)

1. Polymorphism through operators:Here single operator,it shows different actions based on input context.
 - for example:"+" operator,it performs addition operation when inputs are integers.As well as it forms string concatenation,when inputs are strings.
 - Ex:3+5=8-->Addition
 - 'code'+'gnan'codegnan-->string concatenation
 - [1,2,3]+[4,5]
 - 3*5=15-->multiplication operation
 - 3*code=code code code--->repetation
2. Polymorphism through functions:Here a function shows different actions based on input context.
   - Ex:len("Hello")=5-->string length
   - len([1,2,3])=3--->list length
3. Plymorphism through methods:
  - Method Overriding:(Through inheritance).A method defined in child class,that is exist in parent class with the same name.whenever if you call that method through child class object,the child class method override the parent class method.This is known as Method Overriding.<br>
    `Syntax:   class A:`<br>
    `            def show():`<br>
    `          class B:`     <br>
    `            def show():`<br>
    `          b=B()`         <br>
    `          b.show()`      <br>

```python
#Polymorphism through methods
#Method Overriding
class Student:
    def __init__(self,name,age,gender):
        self.name=name
        self.age=age
        self.gender=gender
    def Std_details(self):
        return f"My name is {self.name} and I am {self.age} years old"
class Employee(Student):
    def __init__(self,name,age,gender,salary):
        super().__init__(name,age,gender)
        self.salary=salary
    def Std_details(self):
        print(super().Std_details())
        return f"My name is {self.name} and My salary is {self.salary}"
e=Employee("Varsha",20,"F",50000)
print(e.Std_details())
```

- `Method Overloading`:A class contains multiple methods with same name with different parameter list,In that case last defined method overrides the previous defined method.This is called as Method Overloading.
   - Directly python dosen't supports method overloading but we can achieve same functionality through default arguments,variable length positional,or variable length keyword arguments.

```python
#Method Overloading
class Math:
    def add(self,a,b):
        return a+b
    def add(self,a,b,c):
        return a+b+c
    def add(self,a,b,c,d):
        return a+b+c+d
m=Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
#Abstraction
from abc import ABC, abstractmethod
class Mobile(ABC):
    @abstractmethod
    def ram(self):
        pass
    @abstractmethod
    def rom(self):
        pass
    @abstractmethod
    def battery(self):
        pass
    #concrete method implementation
    def display(self,Hzs):
        return f" This mobile display quality is {Hzs} Hzs"
class Oppo(Mobile):
    def __init__(self,model,battery_size,display_quality):
        self.model=model
        #self.ram=ram
        #self.rom=rom
        self.battery_size=battery_size
        self.display_quality=display_quality
    def ram(self,ram_size):
        return f"The Oppo mobile {self.model} has {ram_size}GB RAM"
    def rom(self,rom_size):
        return f"The Oppo mobile {self.model} has {rom_size}GB ROM"
    def battery(self):
        return f"The Oppo mobile {self.model} has {self.battery_size} mah Battery Capacity"
    def camera(self,primary,secondary):
        return f"The Oppo mobile {self.model} have primary camera with {primary}MP and secondary camera with {secondary}MP"
#m=Mobile()    TypeError: Can't instantiate abstract class Mobile without an implementation for abstract methods 'battery', 'ram', 'rom'
f23=Oppo("f23",5000,120)
print(f23.ram(8))
print(f23.rom(256))
print(f23.battery())
print(f23.camera(32,64))
print(f23.display(120))
```

```python
#find the perimeter,surface area,volume of 2d,3d shapes.class names as cube,sphere,cone for 2D and square,rectangle,triangle for 3D
#2D
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def perimeter(self):
        pass
    @abstractmethod
    def area(self):
        pass
class Square(Shape):
    def __init__(self,side):
        self.side=side
    def perimeter(self):
        return 4*self.side
    def area(self):
        return self.side*self.side
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def perimeter(self):
        return 2*(self.length+self.breadth)
    def area(self):
        return self.length*self.breadth
class Triangle(Shape):
    def __init__(self,base,height,side1,side2):
        self.base=base
        self.height=height
        self.side1=side1
        self.side2=side2
    def perimeter(self):
        return (self.base+self.side1+self.side2)
    def area(self):
        return 0.5*self.base*self.height
s=Square(4)
print("Square")
print("Perimeter:",s.perimeter())
print("Area:",s.area())
r=Rectangle(6,3)
print("Rectangle")
print("Perimeter:",r.perimeter())
print("Area:",r.area())
t=Triangle(6,4,5,5)
print("Triangle:",t.perimeter())
print("Area:",t.area())
```

```python
#3D
from abc import ABC, abstractmethod
from math import pi
class Shape(ABC):
    @abstractmethod
    def surface_area(self):
        pass
    @abstractmethod
    def volume(self):
        pass
class Cube(Shape):
    def __init__(self, side):
        self.side=side
    def surface_area(self):
        return 6*self.side*self.side
    def volume(self):
        return self.side**3
class Sphere(Shape):
    def __init__(self,radius):
        self.radius=radius
    def surface_area(self):
        return 4*pi*self.radius*self.radius
    def volume(self):
        return (4/3)*pi*self.radius**3
class Cone(Shape):
    def __init__(self, radius, height):
        self.radius=radius
        self.height=height
    def surface_area(self):
        slant=0.5**(self.radius**2+self.height**2)          #slant height=sqrt(r**2+h**2)
        return pi*self.radius*(self.radius+slant)           #surface area=pi*r*(r+slant height)
    def volume(self):
        return (1/3)*pi*self.radius**2*self.height
c=Cube(5)
print("Cube")
print("Surface Area:",c.surface_area())
print("Volume:",c.volume())
sp=Sphere(3)
print("Sphere")
print("Surface Area:",round(sp.surface_area(),2))
print("Volume:",round(sp.volume(),2))
c=Cone(3,4)
print("Cone")
print("Surface Area:",round(c.surface_area(),2))
print("Volume:",round(c.volume(),2))
```

```python
class Person:
    def __init__(self,name,age,gender):
        try:
            if not name.isalpha():
                raise ValueError("Name should contain only letters.")
            if age<=0:
                raise ValueError("Age must be greater than 0.")
            if gender.lower() not in ["male","female"]:
                raise ValueError("Gender should be Male or Female.")
            self.name=name
            self.age=age
            self.gender=gender
        except ValueError as e:
            print("Value Error:", e)
    def details(self):
        try:
            print("Name:",self.name)
            print("Age:",self.age)
            print("Gender:",self.gender)
        except AttributeError:
            print("Person details are not available because of invalid input.")
p1=Person("Varsha",21,"Female")
p1.details()
```
