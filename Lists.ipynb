### List Characteristics

```python
#Define list
lst=[1,2,4.5,"Hi","Hello",1,3,4]
print(lst)
```

```python
#List allows indexing and slicing
#Accessing list items
lst=[1,2,4.5,"Hi","Hello",1,3,4]
print(lst[3])
print(lst[-5])
print(lst[::-1])
```

```python
lst=[10,20,30,40,50,60,70]
print(lst[:5:-1])
print(lst[:500])
print(lst[::-1])
print(lst[-4:10])
```

```python
#Empty list creation
l1=[]
l2=list()
print(type(l1), type(l2))
```

```python
#List operations
#1.List Concatination (using + operator)
#2.List Repetation (using * operator)
l1=[1,2,3]
l2=[3,4,5,6]
print(l1+l2)
print(l1+l1)
print(l1*3)
print([0] * 10)
```

### List Methods
- Item adding Methods
  - append(item) : It adds item at end of list
  - extend(iterable) : It adds all the iterator elements at end of the list
  - insert(pos, item) : It adds item at specified position

```python
lst=[10,20,30,40]
lst.append(50)
print(lst)
lst.extend([1,2,3])
print(lst)
lst.append([100,200])       #list in list
print(lst)
lst.insert(0,100)
print(lst)
```

### Item Removing Methods
- remove(item) : It removes the first occurence item from the list if it exists in list, otherwise 'Value Error'
- pop(pos) :It removes and return the specified position item if position is passed,otherwise it remove and returns the last item from list
- clear() : It removes all elements from list

```python
lst=[10,20,30,20,40]
lst.remove(20)
print(lst)
```

```python
lst=[10,20,30,20,40,50,20]
val=lst.pop()
print(val)
print(lst)
val2=lst.pop(0) #It removes the 0th position
print(val2)
print(lst)
lst.clear()
print(lst)
```

# Index and Count()
- index(item) : If item is present in list it returns first occurence of the index,otherwise 'Value Error'
- count(item) : It return the count of total occurence of item

```python
lst=[10,20,30,20,40,50]
ind=lst.index(20)
print(ind)
print(lst.count(20))
```

## Sorting items
- sort(reverse=False) : By default it sorts the original list into ascending order.If reverse is True,it sorts in descending order

```python
lst=[1,4,2,3,5,-5,3,15,-90,30]
lst.sort()
print(lst)
lst.sort(reverse=True)
print(lst)
```

```python
lst=['Banty','Sunny','Ammu','Naini']
lst.sort()
print(lst)
lst.sort(reverse=True)
print(lst)
```

```python
#Built-in Functions : len(),min(),max(),sum()
lst=[1,2,3,4,5]
print(len(lst),min(lst),max(lst),sum(lst))
print("Average: ",sum(lst)/len(lst))
```
