### Set Characteristics

```python
#Set unordered collection
#Set return unique elements
s={1,2,3,2,3,'hi','a',2.02,(10,20,30)}
print(s)
#print(s[0]) # set does not supports indexing
```

```python
#list can't be used as set element
s={1,2,[3,2,3],'hi','a',2.02,(10,20,30)}
print(s)
```

### Set Methods

```python
#Elements adding method
#add(ele):It adds the element to set if element not in set,otherwise nothing
s={1,2,3,4,5,4,3}
print(s)
s.add(100)
s.add(5)
s.add("Code")
print(s)
```

## Element removing methods
- remove(item):It removes the item if it exists in set,otherwise 'Key Error'
- discard(item):It removes the item if it exists in set,otherwise nothing

```python
s={1,2,3,4,5}
s.remove(5)
print(s)
#s.remove(100) # Key Error
s.discard(3)
print(s)
s.discard(100)
print(s)
```

## Sets Operations
- union(set): It returns all the unique elements from all sets

```python
s1={1,2,3}
s2={3,4,5}
s3={1,2,8,9}
union=s1.union(s2)
print(union)
union=s1.union(s2,s3)
print(union)
res=s1 | s2 | s3
print(res)
```

- intersection(set): It returns the common elements from all sets

```python
s1={1,2,3,5}
s2={1,2,3,4,5,6}
s3={1,2,8,9}
s4={1,2}
union=s1.intersection(s2)
print(union)
union=s1.intersection(s2,s3)
print(union)
res=s1 & s2 & s3 & s4
print(res)
```

- difference(set):It returns a set of elements those are not in another set

```python
s1={1,3,5,7}
s2={4,5,6,7}
s3={5,1,2,8,9}
d=s1.difference(s2)
print(d)
d=s1.difference(s2,s3)
print(d)
res=s1 - s2 - s3
print(res)
res=(s1 | s2) - s3
print(res)
```

- symmetric_difference(set): It returns the non common elements of all sets

```python
s1={1,3,5,7}
s2={4,5,6,7}
s3={5,1,2,8,9}
sy_diff=s1.symmetric_difference(s2)
print(sy_diff)
res=s1 ^ s2 
print(res)
res=s1 ^ s2 ^ s3
print(res)
```

```python
#issubset(set)
s1={1,2,3,4,5}
s2={3,4,5}
subset=s2.issubset(s1)
print(subset)
```

```python
#issuperset(set)
s1={3,4,5}
s2={1,2,3,4,5}
superset=s2.issuperset(s1)
print(superset)
```

```python
#isdisjoint(set)
s1={1,2,3}
s2={6,4,5}
d=s2.isdisjoint(s1)
print(d)
```
