### Recursions

```python
def fun(x):
    if x==0:
        return 0
    x=x-1
    fun(x)
    print(x)
fun(3)
```

```python
def fun(x):
    if x==0:
        return 0
    fun(x-1)
    print(x)
fun(3)
```

```python
#print 1 to n using recursion
def fun(x):
    if x==0:
        return 0
    print(x)
    fun(x-1)
fun(3)
```

```python
#factorial
def factorial(x):
    if x==0 or x==1:
        return 1
    return x*factorial(x-1)
print(factorial(5))
```

```python
#find the sum of digits in a number
def sum(x):
    if x==0:
        return 0
    return x%10+sum(x//10)
print(sum(456))
```

```python
#count digits in a number
def sum(x):
    if x==0:
        return 0
    return 1+sum(x//10)
print(sum(456))
print(sum(1257678345))
```

### Lambda Functions

```python
#find the square of a number using lambda function
s=lambda x:x**2
print(s(20))
```

```python
#find the squares of list of integers using map and lambda
arr=[1,2,3,4,5]
print(list(map(lambda x:x**2,arr)))
```

```python
#even numbers
arr=[1,2,3,4,56,55]
print(list(filter(lambda x:x%2==0,arr)))
```

```python
n=int(input())
res='Even' if n%2==0 else 'Odd'
print(res)
```

```python
#check whether the number is even or odd using lambda
arr=[1,2,3,4,56,55]
print(list(map(lambda x:'Even' if x%2==0 else 'Odd',arr)))
```

```python
n=lambda x:'Even' if x%2==0 else 'Odd'
print(n(10))
```

```python
arr=[1,2,3,4,56,55]
print(list(map(lambda x:f"{x} - Even" if x%2==0 else f"{x} - Odd",arr)))
```

```python
#Find the frequency of each character
str=input()
count={}
for char in str:
    if char in count:
        count[char]+=1
    else:
        count[char]=1
print(count)
```

```python
str=input()
count={}
for char in str:
        count[char]=count.get(char,0)+1
print(count)
```
